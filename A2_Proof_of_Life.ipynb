# Engineering Problem

The engineering problem investigated in this assignment is the relationship
between concrete age and compressive strength.

The aim is to use the available concrete data to investigate whether
compressive strength increases as the age of concrete increases.

```python
import matplotlib.pyplot as plt

plt.scatter(data["Age_days"], data["Compressive_Strength_MPa"])

plt.xlabel("Concrete Age (days)")
plt.ylabel("Compressive Strength (MPa)")
plt.title("Concrete Age vs Compressive Strength")

plt.show()
```

```python
average_strength = data.groupby("Age_days")["Compressive_Strength_MPa"].mean()

print(average_strength)
```

## Results and Conclusion

The analysis shows that the average compressive strength increased as
the concrete age increased.

At 7 days, the average compressive strength was 25.15 MPa.
At 14 days, it increased to 31.35 MPa.
At 28 days, it increased to 42.14 MPa.

Therefore, the data in this sample shows a positive relationship between
concrete age and compressive strength. The concrete had a higher average
compressive strength at 28 days than at 7 days.

```python
average_strength.plot(kind="bar")

plt.xlabel("Concrete Age (days)")
plt.ylabel("Average Compressive Strength (MPa)")
plt.title("Average Compressive Strength at Different Ages")

plt.show()
```
